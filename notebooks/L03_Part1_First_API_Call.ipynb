# Lesson 03: Part 1 — First LLM API Call

### What is an API?
In web chat (like ChatGPT or Gemini web), you type messages into a browser.

In software engineering, programs connect directly to the model using an **API** (Application Programming Interface). This notebook shows how to make that call in Python.

---
## Step 1: Get Your Free API Key
1. Open [Google AI Studio](https://aistudio.google.com/).
2. Sign in with your Google account.
3. Click **Get API key** -> **Create API key**.
4. Copy your key. (Free tier: 15 requests per minute at $0 cost).

---
## Step 2: Save Key in Colab Secrets
Never write your secret key directly in Python code.

1. Click the **Key icon (Secrets)** on the left sidebar.
2. Click **Add new secret**.
3. Name: `GEMINI_API_KEY`.
4. Value: Paste your key from Google AI Studio.
5. Turn on **Notebook access**.

```python
# Install official Google GenAI library (~5 seconds)
!pip install --quiet -U google-genai

# Connect to the model using your secret key
from google.colab import userdata
from google import genai

api_key = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=api_key)
print("Connected to Gemini API successfully!")
```

---
## Step 3: Send Your First Request
We send a simple question to the `gemini-3.5-flash-lite` model.

```python
# Send a question to Gemini 3.5 Flash
response = client.models.generate_content(
    model='gemini-3.5-flash-lite',
    contents='Explain what an API is in one simple sentence.'
)

print("Model Response:")
print(response.text.strip())
```

---
## Step 4: Token Usage (Digital Receipt)
Every API call reports the exact number of tokens used:
- **Prompt tokens:** Number of tokens in your question.
- **Candidate tokens:** Number of tokens in the answer.

```python
# Inspect token usage
print(f"Prompt tokens:    {response.usage_metadata.prompt_token_count}")
print(f"Candidate tokens: {response.usage_metadata.candidates_token_count}")
print(f"Total tokens:     {response.usage_metadata.total_token_count}")
```

---
## Step 5: Try It Yourself
Change `my_question` below and run this cell.

```python
# Type any question you want:
my_question = "What is the capital of Uzbekistan?"

my_response = client.models.generate_content(
    model='gemini-3.5-flash-lite',
    contents=my_question
)

print(f"Question: {my_question}")
print(f"Answer: {my_response.text.strip()}\n")
print(f"Tokens used: {my_response.usage_metadata.total_token_count}")
```
